```python
covid = precio_diario_95[
    (precio_diario_95["fecha"] >= "2020-03-01") &
    (precio_diario_95["fecha"] <= "2020-06-30")
].copy()

ucrania = precio_diario_95[
    (precio_diario_95["fecha"] >= "2022-01-01") &
    (precio_diario_95["fecha"] <= "2022-06-30")
].copy()

print("COVID-19")
print(covid.head())
print(covid.tail())

print("\nRusia-Ucrania")
print(ucrania.head())
print(ucrania.tail())
```

```python
# COVID-19
covid_antes = precio_diario_95[
    (precio_diario_95["fecha"] >= "2020-01-01") &
    (precio_diario_95["fecha"] < "2020-03-14")
]["pvp_eur_litro"].mean()

covid_durante = precio_diario_95[
    (precio_diario_95["fecha"] >= "2020-03-14") &
    (precio_diario_95["fecha"] <= "2020-06-07")
]["pvp_eur_litro"].mean()


# Rusia-Ucrania
ucrania_antes = precio_diario_95[
    (precio_diario_95["fecha"] >= "2022-01-01") &
    (precio_diario_95["fecha"] < "2022-02-24")
]["pvp_eur_litro"].mean()

ucrania_durante = precio_diario_95[
    (precio_diario_95["fecha"] >= "2022-02-24") &
    (precio_diario_95["fecha"] <= "2022-12-31")
]["pvp_eur_litro"].mean()


print("COVID-19")
print("Media antes:", round(covid_antes, 3), "€/L")
print("Media durante:", round(covid_durante, 3), "€/L")
print("Diferencia:", round(covid_durante - covid_antes, 3), "€/L")

print("\nRusia-Ucrania")
print("Media antes:", round(ucrania_antes, 3), "€/L")
print("Media durante:", round(ucrania_durante, 3), "€/L")
print("Diferencia:", round(ucrania_durante - ucrania_antes, 3), "€/L")
```

```python
resumen_shocks = pd.DataFrame({
    "shock": ["COVID-19", "Rusia-Ucrania"],
    "precio_antes": [covid_antes, ucrania_antes],
    "precio_durante": [covid_durante, ucrania_durante],
})

resumen_shocks["diferencia_eur_litro"] = (
    resumen_shocks["precio_durante"]
    - resumen_shocks["precio_antes"]
)

resumen_shocks["diferencia_porcentaje"] = (
    resumen_shocks["diferencia_eur_litro"]
    / resumen_shocks["precio_antes"]
) * 100

print(resumen_shocks.round(3))
```

```python
# ============================================================
# PRUEBA FINAL: añadir gasolina_lag_1 y gasolina_lag_7
# ============================================================

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Variables del modelo actual
# ------------------------------------------------------------

variables_actuales = [
    "brent_eur_barril",
    "brent_eur_lag_1",
    "brent_eur_lag_3",
    "brent_eur_lag_7",
    "brent_cambio_1d",
    "brent_cambio_7d",
    "mes",
    "día_semana",
    "dias_desde_inicio"
]

# ------------------------------------------------------------
# 2. Variables del modelo ampliado
# ------------------------------------------------------------

variables_ampliadas = variables_actuales + [
    "gasolina_lag_1",
    "gasolina_lag_7"
]

# ------------------------------------------------------------
# 3. Comprobar que las variables existen
# ------------------------------------------------------------

print("Variables que faltan:")

faltantes = [
    col for col in variables_ampliadas
    if col not in train_val.columns
]

print(faltantes)

# ------------------------------------------------------------
# 4. Eliminar posibles NaN de las nuevas variables
#    Solo afecta al entrenamiento, NO al test 2025
# ------------------------------------------------------------

train_val_ampliado = train_val.dropna(
    subset=["gasolina_lag_1", "gasolina_lag_7"]
).copy()

print("\nTrain ampliado:", train_val_ampliado.shape)

# ------------------------------------------------------------
# 5. Entrenar modelo ampliado
#    Entrenamiento: 2016-2023
#    Validación: 2024
# ------------------------------------------------------------

modelo_ampliado = LinearRegression()

modelo_ampliado.fit(
    train_val_ampliado[variables_ampliadas],
    train_val_ampliado["pvp_eur_litro"]
)

pred_2024_ampliado = modelo_ampliado.predict(
    val[variables_ampliadas]
)

# ------------------------------------------------------------
# 6. Métricas del modelo ampliado en 2024
# ------------------------------------------------------------

mae_2024_ampliado = mean_absolute_error(
    val["pvp_eur_litro"],
    pred_2024_ampliado
)

rmse_2024_ampliado = np.sqrt(
    mean_squared_error(
        val["pvp_eur_litro"],
        pred_2024_ampliado
    )
)

r2_2024_ampliado = r2_score(
    val["pvp_eur_litro"],
    pred_2024_ampliado
)

# ------------------------------------------------------------
# 7. Mostrar comparación
# ------------------------------------------------------------

print("\n==============================")
print("COMPARACIÓN EN VALIDACIÓN 2024")
print("==============================")

print("\nMODELO ACTUAL")
print("MAE :", round(mae_2024, 6))
print("RMSE:", round(rmse_2024, 6))
print("R²  :", round(r2_2024, 6))

print("\nMODELO AMPLIADO")
print("MAE :", round(mae_2024_ampliado, 6))
print("RMSE:", round(rmse_2024_ampliado, 6))
print("R²  :", round(r2_2024_ampliado, 6))

# ------------------------------------------------------------
# 8. Diferencia respecto al modelo actual
# ------------------------------------------------------------

print("\n==============================")
print("CAMBIO")
print("==============================")

print(
    "Cambio MAE :",
    round(mae_2024_ampliado - mae_2024, 6)
)

print(
    "Cambio RMSE:",
    round(rmse_2024_ampliado - rmse_2024, 6)
)

print(
    "Cambio R²  :",
    round(r2_2024_ampliado - r2_2024, 6)
)

# ------------------------------------------------------------
# 9. Si mejora, entrenamos el modelo final con 2016-2024
#    y evaluamos 2025
# ------------------------------------------------------------

mejora = (
    mae_2024_ampliado < mae_2024
    and
    rmse_2024_ampliado < rmse_2024
)

if mejora:

    print("\n>>> El modelo ampliado mejora en 2024.")
    print(">>> Se utilizará para el test final de 2025.")

    # Entrenamiento final: 2016-2024
    train_2025_ampliado = train_2025.dropna(
        subset=["gasolina_lag_1", "gasolina_lag_7"]
    ).copy()

    modelo_final_ampliado = LinearRegression()

    modelo_final_ampliado.fit(
        train_2025_ampliado[variables_ampliadas],
        train_2025_ampliado["pvp_eur_litro"]
    )

    # Predicción 2025
    pred_2025_ampliado = modelo_final_ampliado.predict(
        val_2025[variables_ampliadas]
    )

    # Métricas finales
    mae_2025_ampliado = mean_absolute_error(
        val_2025["pvp_eur_litro"],
        pred_2025_ampliado
    )

    rmse_2025_ampliado = np.sqrt(
        mean_squared_error(
            val_2025["pvp_eur_litro"],
            pred_2025_ampliado
        )
    )

    r2_2025_ampliado = r2_score(
        val_2025["pvp_eur_litro"],
        pred_2025_ampliado
    )

    print("\n==============================")
    print("TEST FINAL 2025")
    print("==============================")

    print("MAE 2025 :", round(mae_2025_ampliado, 6))
    print("RMSE 2025:", round(rmse_2025_ampliado, 6))
    print("R² 2025  :", round(r2_2025_ampliado, 6))

else:

    print("\n>>> El modelo ampliado NO mejora suficientemente.")
    print(">>> Mantenemos el modelo actual.")
```

```python
# ============================================================
# MODELO PREDICTIVO REAL — PRECIO DE GASOLINA
# ============================================================
#
# Objetivo:
# Predecir el precio de gasolina de un día utilizando
# únicamente información disponible hasta el día anterior.
#
# Train      -> 2016-2023
# Validación -> 2024
# Test final -> 2025
# ============================================================

import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ============================================================
# 1. CREAR DATAFRAME PARA EL MODELO PREDICTIVO
# ============================================================

df_pred = gasolina_brent_eurusd_model.copy()

df_pred = df_pred.sort_values("fecha").reset_index(drop=True)


# ============================================================
# 2. CREAR VARIABLES DISPONIBLES ANTES DEL DÍA A PREDECIR
# ============================================================

# Brent del día anterior
df_pred["brent_eur_lag_1_pred"] = (
    df_pred["brent_eur_barril"].shift(1)
)

# Brent de 2 días antes
df_pred["brent_eur_lag_2_pred"] = (
    df_pred["brent_eur_barril"].shift(2)
)

# Brent de 4 días antes
df_pred["brent_eur_lag_4_pred"] = (
    df_pred["brent_eur_barril"].shift(4)
)

# Brent de 8 días antes
df_pred["brent_eur_lag_8_pred"] = (
    df_pred["brent_eur_barril"].shift(8)
)

# Cambio del Brent de los últimos días,
# desplazado un día para no utilizar información del día objetivo
df_pred["brent_cambio_1d_pred"] = (
    df_pred["brent_cambio_1d"].shift(1)
)

df_pred["brent_cambio_7d_pred"] = (
    df_pred["brent_cambio_7d"].shift(1)
)

# Variables temporales conocidas de antemano
df_pred["mes_pred"] = df_pred["mes"]

df_pred["día_semana_pred"] = df_pred["día_semana"]

df_pred["dias_desde_inicio_pred"] = (
    df_pred["dias_desde_inicio"]
)


# ============================================================
# 3. PRECIO DE GASOLINA HISTÓRICO
# ============================================================
#
# Estas variables sí son válidas para una predicción real:
# el precio de ayer y el de hace 7 días ya se conocen.
# ============================================================

df_pred["gasolina_lag_1_pred"] = (
    df_pred["pvp_eur_litro"].shift(1)
)

df_pred["gasolina_lag_7_pred"] = (
    df_pred["pvp_eur_litro"].shift(7)
)


# ============================================================
# 4. VARIABLES FINALES
# ============================================================

variables_pred = [
    "brent_eur_lag_1_pred",
    "brent_eur_lag_2_pred",
    "brent_eur_lag_4_pred",
    "brent_eur_lag_8_pred",
    "brent_cambio_1d_pred",
    "brent_cambio_7d_pred",
    "gasolina_lag_1_pred",
    "gasolina_lag_7_pred",
    "mes_pred",
    "día_semana_pred",
    "dias_desde_inicio_pred"
]


# ============================================================
# 5. ELIMINAR FILAS CON NaN
# ============================================================

df_pred_model = df_pred.dropna(
    subset=variables_pred + ["pvp_eur_litro"]
).copy()


print("Filas disponibles:", len(df_pred_model))
print("Variables:", variables_pred)


# ============================================================
# 6. TRAIN / VALIDACIÓN / TEST
# ============================================================

train_pred = df_pred_model[
    df_pred_model["fecha"].dt.year < 2024
].copy()

val_pred = df_pred_model[
    df_pred_model["fecha"].dt.year == 2024
].copy()

test_pred = df_pred_model[
    df_pred_model["fecha"].dt.year == 2025
].copy()


print("\n==============================")
print("DIVISIÓN DE DATOS")
print("==============================")

print("Train:", train_pred.shape)
print("Validación:", val_pred.shape)
print("Test:", test_pred.shape)


# ============================================================
# 7. MODELO BASE
# ============================================================

modelo_pred = LinearRegression()

modelo_pred.fit(
    train_pred[variables_pred],
    train_pred["pvp_eur_litro"]
)


# ============================================================
# 8. VALIDACIÓN 2024
# ============================================================

pred_2024_pred = modelo_pred.predict(
    val_pred[variables_pred]
)

mae_2024_pred = mean_absolute_error(
    val_pred["pvp_eur_litro"],
    pred_2024_pred
)

rmse_2024_pred = np.sqrt(
    mean_squared_error(
        val_pred["pvp_eur_litro"],
        pred_2024_pred
    )
)

r2_2024_pred = r2_score(
    val_pred["pvp_eur_litro"],
    pred_2024_pred
)


print("\n==============================")
print("VALIDACIÓN 2024")
print("==============================")

print("MAE :", round(mae_2024_pred, 6))
print("RMSE:", round(rmse_2024_pred, 6))
print("R²  :", round(r2_2024_pred, 6))


# ============================================================
# 9. REENTRENAMIENTO FINAL
# ============================================================
#
# Una vez validado el modelo, incorporamos 2024 al entrenamiento.
# 2025 permanece completamente separado.
# ============================================================

train_final_pred = df_pred_model[
    df_pred_model["fecha"].dt.year < 2025
].copy()


modelo_final_pred = LinearRegression()

modelo_final_pred.fit(
    train_final_pred[variables_pred],
    train_final_pred["pvp_eur_litro"]
)


# ============================================================
# 10. TEST FINAL 2025
# ============================================================

pred_2025_pred = modelo_final_pred.predict(
    test_pred[variables_pred]
)


mae_2025_pred = mean_absolute_error(
    test_pred["pvp_eur_litro"],
    pred_2025_pred
)

rmse_2025_pred = np.sqrt(
    mean_squared_error(
        test_pred["pvp_eur_litro"],
        pred_2025_pred
    )
)

r2_2025_pred = r2_score(
    test_pred["pvp_eur_litro"],
    pred_2025_pred
)


print("\n==============================")
print("TEST FINAL 2025")
print("==============================")

print("MAE :", round(mae_2025_pred, 6))
print("RMSE:", round(rmse_2025_pred, 6))
print("R²  :", round(r2_2025_pred, 6))


# ============================================================
# 11. RESULTADOS DÍA A DÍA DE 2025
# ============================================================

resultado_2025_pred = test_pred[
    ["fecha", "pvp_eur_litro"]
].copy()

resultado_2025_pred["prediccion"] = pred_2025_pred

resultado_2025_pred["error"] = (
    resultado_2025_pred["pvp_eur_litro"]
    - resultado_2025_pred["prediccion"]
)

resultado_2025_pred["error_abs"] = (
    resultado_2025_pred["error"].abs()
)


# ============================================================
# 12. ERROR MEDIO RELATIVO
# ============================================================

precio_medio_2025_pred = (
    test_pred["pvp_eur_litro"].mean()
)

error_porcentual_2025_pred = (
    mae_2025_pred /
    precio_medio_2025_pred
) * 100


print("\n==============================")
print("ERROR RELATIVO 2025")
print("==============================")

print(
    "Precio medio 2025:",
    round(precio_medio_2025_pred, 6)
)

print(
    "MAE:",
    round(mae_2025_pred, 6)
)

print(
    "Error medio porcentual:",
    round(error_porcentual_2025_pred, 4),
    "%"
)


# ============================================================
# 13. SESGO DEL MODELO
# ============================================================

error_medio_2025_pred = (
    resultado_2025_pred["error"].mean()
)

dias_sobreestima_pred = (
    resultado_2025_pred["error"] < 0
).sum()

dias_subestima_pred = (
    resultado_2025_pred["error"] > 0
).sum()


print("\n==============================")
print("SESGO 2025")
print("==============================")

print(
    "Error medio:",
    round(error_medio_2025_pred, 6)
)

print(
    "Días sobreestimados:",
    dias_sobreestima_pred
)

print(
    "Días subestimados:",
    dias_subestima_pred
)


# ============================================================
# 14. 10 MAYORES ERRORES
# ============================================================

print("\n==============================")
print("10 MAYORES ERRORES — 2025")
print("==============================")

print(
    resultado_2025_pred
    .sort_values("error_abs", ascending=False)
    .head(10)
)


# ============================================================
# 15. ERROR POR MES
# ============================================================

error_mensual_2025_pred = (
    resultado_2025_pred
    .groupby(
        resultado_2025_pred["fecha"].dt.month
    )["error_abs"]
    .agg(["mean", "max", "count"])
)


print("\n==============================")
print("ERROR MENSUAL — 2025")
print("==============================")

print(error_mensual_2025_pred)


# ============================================================
# 16. RESUMEN FINAL
# ============================================================

resumen_modelo_predictivo = pd.DataFrame({
    "Periodo": [
        "Validación 2024",
        "Test final 2025"
    ],
    "MAE": [
        mae_2024_pred,
        mae_2025_pred
    ],
    "RMSE": [
        rmse_2024_pred,
        rmse_2025_pred
    ],
    "R²": [
        r2_2024_pred,
        r2_2025_pred
    ]
})


print("\n==============================")
print("RESUMEN FINAL")
print("==============================")

print(
    resumen_modelo_predictivo.round(6)
)
```

```python
# ============================================================
# MODELO PREDICTIVO ESTRICTO — SIN PRECIO DE GASOLINA
# ============================================================
#
# Objetivo:
# Predecir el precio de gasolina del día siguiente utilizando
# únicamente información disponible antes del día objetivo.
#
# NO utilizamos:
# - pvp del propio día objetivo
# - gasolina_lag_1
# - gasolina_lag_7
#
# Sí utilizamos:
# - Brent histórico
# - cambios históricos del Brent
# - variables temporales
#
# Train      -> 2016-2023
# Validación -> 2024
# Test       -> 2025
# ============================================================

import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ============================================================
# 1. COPIA Y ORDEN TEMPORAL
# ============================================================

df_estricto = gasolina_brent_eurusd_model.copy()

df_estricto = df_estricto.sort_values(
    "fecha"
).reset_index(drop=True)


# ============================================================
# 2. VARIABLES DE BRENT DISPONIBLES ANTES DEL DÍA OBJETIVO
# ============================================================

# Brent del día anterior
df_estricto["brent_lag_1"] = (
    df_estricto["brent_eur_barril"].shift(1)
)

# Brent de 2 días antes
df_estricto["brent_lag_2"] = (
    df_estricto["brent_eur_barril"].shift(2)
)

# Brent de 3 días antes
df_estricto["brent_lag_3"] = (
    df_estricto["brent_eur_barril"].shift(3)
)

# Brent de 7 días antes
df_estricto["brent_lag_7"] = (
    df_estricto["brent_eur_barril"].shift(7)
)


# ============================================================
# 3. CAMBIOS DEL BRENT
# ============================================================

# Cambio de Brent de hace 1 día
df_estricto["brent_cambio_1d"] = (
    df_estricto["brent_cambio_1d"].shift(1)
)

# Cambio de Brent de hace 7 días
df_estricto["brent_cambio_7d"] = (
    df_estricto["brent_cambio_7d"].shift(1)
)


# ============================================================
# 4. VARIABLES TEMPORALES
# ============================================================

df_estricto["mes_modelo"] = (
    df_estricto["mes"]
)

df_estricto["dia_semana_modelo"] = (
    df_estricto["día_semana"]
)

df_estricto["dias_inicio_modelo"] = (
    df_estricto["dias_desde_inicio"]
)


# ============================================================
# 5. DEFINIR VARIABLES DEL MODELO
# ============================================================

variables_estricto = [
    "brent_lag_1",
    "brent_lag_2",
    "brent_lag_3",
    "brent_lag_7",
    "brent_cambio_1d",
    "brent_cambio_7d",
    "mes_modelo",
    "dia_semana_modelo",
    "dias_inicio_modelo"
]


print("Variables utilizadas:")
print(variables_estricto)


# ============================================================
# 6. CREAR OBJETIVO
# ============================================================
#
# La fila de hoy utiliza información conocida hasta hoy,
# pero el objetivo es el precio de MAÑANA.
#
# Por eso desplazamos el precio un día hacia arriba.
# ============================================================

df_estricto["objetivo_dia_siguiente"] = (
    df_estricto["pvp_eur_litro"].shift(-1)
)

# Fecha correspondiente al precio que queremos predecir
df_estricto["fecha_objetivo"] = (
    df_estricto["fecha"].shift(-1)
)


# ============================================================
# 7. ELIMINAR FILAS INCOMPLETAS
# ============================================================

df_estricto_model = df_estricto.dropna(
    subset=variables_estricto
    + [
        "objetivo_dia_siguiente",
        "fecha_objetivo"
    ]
).copy()


print("\nFilas disponibles:")
print(len(df_estricto_model))


# ============================================================
# 8. DIVISIÓN TEMPORAL
# ============================================================
#
# IMPORTANTE:
# dividimos por el AÑO DEL OBJETIVO,
# no por el año de la fila de origen.
# ============================================================

train_estricto = df_estricto_model[
    df_estricto_model["fecha_objetivo"].dt.year < 2024
].copy()

val_estricto = df_estricto_model[
    df_estricto_model["fecha_objetivo"].dt.year == 2024
].copy()

test_estricto = df_estricto_model[
    df_estricto_model["fecha_objetivo"].dt.year == 2025
].copy()


print("\n==============================")
print("DIVISIÓN TEMPORAL")
print("==============================")

print(
    "Train:",
    train_estricto.shape
)

print(
    "Validación:",
    val_estricto.shape
)

print(
    "Test:",
    test_estricto.shape
)


# ============================================================
# 9. ENTRENAMIENTO
# ============================================================

modelo_estricto = LinearRegression()

modelo_estricto.fit(
    train_estricto[variables_estricto],
    train_estricto["objetivo_dia_siguiente"]
)


# ============================================================
# 10. VALIDACIÓN 2024
# ============================================================

pred_2024_estricto = modelo_estricto.predict(
    val_estricto[variables_estricto]
)


mae_2024_estricto = mean_absolute_error(
    val_estricto["objetivo_dia_siguiente"],
    pred_2024_estricto
)

rmse_2024_estricto = np.sqrt(
    mean_squared_error(
        val_estricto["objetivo_dia_siguiente"],
        pred_2024_estricto
    )
)

r2_2024_estricto = r2_score(
    val_estricto["objetivo_dia_siguiente"],
    pred_2024_estricto
)


print("\n==============================")
print("VALIDACIÓN 2024")
print("==============================")

print(
    "MAE :",
    round(mae_2024_estricto, 6)
)

print(
    "RMSE:",
    round(rmse_2024_estricto, 6)
)

print(
    "R²  :",
    round(r2_2024_estricto, 6)
)


# ============================================================
# 11. ENTRENAMIENTO FINAL
# ============================================================
#
# Incorporamos 2024 al entrenamiento.
# 2025 permanece completamente separado.
# ============================================================

train_final_estricto = df_estricto_model[
    df_estricto_model["fecha_objetivo"].dt.year < 2025
].copy()


modelo_final_estricto = LinearRegression()

modelo_final_estricto.fit(
    train_final_estricto[variables_estricto],
    train_final_estricto["objetivo_dia_siguiente"]
)


# ============================================================
# 12. TEST FINAL 2025
# ============================================================

pred_2025_estricto = modelo_final_estricto.predict(
    test_estricto[variables_estricto]
)


mae_2025_estricto = mean_absolute_error(
    test_estricto["objetivo_dia_siguiente"],
    pred_2025_estricto
)

rmse_2025_estricto = np.sqrt(
    mean_squared_error(
        test_estricto["objetivo_dia_siguiente"],
        pred_2025_estricto
    )
)

r2_2025_estricto = r2_score(
    test_estricto["objetivo_dia_siguiente"],
    pred_2025_estricto
)


print("\n==============================")
print("TEST FINAL 2025")
print("==============================")

print(
    "MAE :",
    round(mae_2025_estricto, 6)
)

print(
    "RMSE:",
    round(rmse_2025_estricto, 6)
)

print(
    "R²  :",
    round(r2_2025_estricto, 6)
)


# ============================================================
# 13. RESULTADOS DÍA A DÍA
# ============================================================

resultado_2025_estricto = test_estricto[
    [
        "fecha_objetivo",
        "objetivo_dia_siguiente"
    ]
].copy()

resultado_2025_estricto = (
    resultado_2025_estricto
    .rename(
        columns={
            "fecha_objetivo": "fecha",
            "objetivo_dia_siguiente": "pvp_real"
        }
    )
)

resultado_2025_estricto["prediccion"] = (
    pred_2025_estricto
)

resultado_2025_estricto["error"] = (
    resultado_2025_estricto["pvp_real"]
    - resultado_2025_estricto["prediccion"]
)

resultado_2025_estricto["error_abs"] = (
    resultado_2025_estricto["error"].abs()
)


# ============================================================
# 14. ERROR RELATIVO
# ============================================================

precio_medio_2025_estricto = (
    resultado_2025_estricto["pvp_real"].mean()
)

error_relativo_2025_estricto = (
    mae_2025_estricto
    / precio_medio_2025_estricto
) * 100


print("\n==============================")
print("ERROR RELATIVO 2025")
print("==============================")

print(
    "Precio medio:",
    round(precio_medio_2025_estricto, 6)
)

print(
    "MAE:",
    round(mae_2025_estricto, 6)
)

print(
    "Error relativo:",
    round(error_relativo_2025_estricto, 4),
    "%"
)


# ============================================================
# 15. SESGO
# ============================================================

error_medio_estricto = (
    resultado_2025_estricto["error"].mean()
)

sobreestimaciones_estricto = (
    resultado_2025_estricto["error"] < 0
).sum()

subestimaciones_estricto = (
    resultado_2025_estricto["error"] > 0
).sum()


print("\n==============================")
print("SESGO 2025")
print("==============================")

print(
    "Error medio:",
    round(error_medio_estricto, 6),
    "€/L"
)

print(
    "Sobreestimaciones:",
    sobreestimaciones_estricto
)

print(
    "Subestimaciones:",
    subestimaciones_estricto
)


# ============================================================
# 16. MAYORES ERRORES
# ============================================================

print("\n==============================")
print("10 MAYORES ERRORES")
print("==============================")

print(
    resultado_2025_estricto
    .sort_values(
        "error_abs",
        ascending=False
    )
    .head(10)
)


# ============================================================
# 17. RESUMEN FINAL
# ============================================================

resumen_estricto = pd.DataFrame({
    "Periodo": [
        "Validación 2024",
        "Test final 2025"
    ],
    "MAE": [
        mae_2024_estricto,
        mae_2025_estricto
    ],
    "RMSE": [
        rmse_2024_estricto,
        rmse_2025_estricto
    ],
    "R²": [
        r2_2024_estricto,
        r2_2025_estricto
    ]
})


print("\n==============================")
print("RESUMEN MODELO ESTRICTO")
print("==============================")

print(
    resumen_estricto.round(6)
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# ==========================================
# 1. RECONSTRUIR LOS DATOS
# ==========================================

df_auto = gasolina_brent_eurusd_model.copy()

df_auto = df_auto.sort_values("fecha").reset_index(drop=True)

# Lags limpios del precio de la gasolina
df_auto["gasolina_lag_1_clean"] = (
    df_auto["pvp_eur_litro"].shift(1)
)

df_auto["gasolina_lag_7_clean"] = (
    df_auto["pvp_eur_litro"].shift(7)
)

# Objetivo: precio del día siguiente
df_auto["objetivo_dia_siguiente"] = (
    df_auto["pvp_eur_litro"].shift(-1)
)

df_auto["fecha_objetivo"] = df_auto["fecha"].shift(-1)

# ==========================================
# 2. VARIABLES DEL MODELO
# ==========================================

variables_auto = [
    "brent_eur_barril",
    "brent_eur_lag_1",
    "brent_eur_lag_3",
    "brent_eur_lag_7",
    "brent_cambio_1d",
    "brent_cambio_7d",
    "mes",
    "día_semana",
    "dias_desde_inicio",
    "gasolina_lag_1_clean",
    "gasolina_lag_7_clean"
]

# Eliminar filas sin datos necesarios
df_auto = df_auto.dropna(
    subset=variables_auto + [
        "objetivo_dia_siguiente",
        "fecha_objetivo"
    ]
).copy()

# ==========================================
# 3. DIVISIÓN TEMPORAL
# ==========================================

train_auto = df_auto[
    df_auto["fecha_objetivo"].dt.year < 2024
].copy()

val_auto = df_auto[
    df_auto["fecha_objetivo"].dt.year == 2024
].copy()

test_auto = df_auto[
    df_auto["fecha_objetivo"].dt.year == 2025
].copy()

print("Entrenamiento:", len(train_auto))
print("Validación 2024:", len(val_auto))
print("Test 2025:", len(test_auto))

# ==========================================
# 4. VALIDACIÓN 2024
# ==========================================

modelo_validacion = LinearRegression()

modelo_validacion.fit(
    train_auto[variables_auto],
    train_auto["objetivo_dia_siguiente"]
)

pred_validacion = modelo_validacion.predict(
    val_auto[variables_auto]
)

print("\n===== VALIDACIÓN 2024 =====")
print(
    "MAE:",
    round(mean_absolute_error(
        val_auto["objetivo_dia_siguiente"],
        pred_validacion
    ), 6)
)

print(
    "RMSE:",
    round(np.sqrt(mean_squared_error(
        val_auto["objetivo_dia_siguiente"],
        pred_validacion
    )), 6)
)

print(
    "R²:",
    round(r2_score(
        val_auto["objetivo_dia_siguiente"],
        pred_validacion
    ), 6)
)

# ==========================================
# 5. MODELO FINAL: ENTRENAR HASTA 2024
# ==========================================

modelo_auto_final = LinearRegression()

train_final_auto = df_auto[
    df_auto["fecha_objetivo"].dt.year < 2025
].copy()

modelo_auto_final.fit(
    train_final_auto[variables_auto],
    train_final_auto["objetivo_dia_siguiente"]
)

# Predecir 2025
pred_auto_2025 = modelo_auto_final.predict(
    test_auto[variables_auto]
)

# ==========================================
# 6. MÉTRICAS TEST 2025
# ==========================================

y_real = test_auto["objetivo_dia_siguiente"]

print("\n===== TEST FINAL 2025 =====")
print("MAE:", round(mean_absolute_error(y_real, pred_auto_2025), 6))
print("RMSE:", round(np.sqrt(mean_squared_error(y_real, pred_auto_2025)), 6))
print("R²:", round(r2_score(y_real, pred_auto_2025), 6))

# ==========================================
# 7. TABLA DE RESULTADOS
# ==========================================

resultado_auto_2025 = pd.DataFrame({
    "fecha": test_auto["fecha_objetivo"].values,
    "pvp_real": y_real.values,
    "prediccion": pred_auto_2025
})

resultado_auto_2025["error"] = (
    resultado_auto_2025["pvp_real"]
    - resultado_auto_2025["prediccion"]
)

resultado_auto_2025["error_abs"] = (
    resultado_auto_2025["error"].abs()
)

display(resultado_auto_2025.head(10))

# ==========================================
# 8. GRÁFICO REAL VS PREDICCIÓN
# ==========================================

plt.figure(figsize=(16, 6))

plt.plot(
    resultado_auto_2025["fecha"],
    resultado_auto_2025["pvp_real"],
    label="Precio real",
    linewidth=2
)

plt.plot(
    resultado_auto_2025["fecha"],
    resultado_auto_2025["prediccion"],
    label="Predicción del modelo",
    linewidth=1.5
)

plt.title("Precio real vs predicción de gasolina — Test 2025")
plt.xlabel("Fecha")
plt.ylabel("Precio (€/L)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
